# Track aircraft in a video

Calls [`infer/track.py`](../infer/track.py) the same way the 2023 notebook did: `persist=True`, `conf=0.20`.

Weights and test mp4s live in `original-data/` (gitignored). Example:

```text
original-data/aircraft_tracking_use/pt/ats4.pt
original-data/aircraft_tracking_use/test_mp4/fighter1.mp4
```

````python
import os
from pathlib import Path

os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
os.chdir(ROOT)

weights = ROOT / "original-data" / "aircraft_tracking_use" / "pt" / "ats4.pt"
source = ROOT / "original-data" / "aircraft_tracking_use" / "test_mp4" / "fighter1.mp4"
weights.exists(), source.exists()
````

````python
import subprocess
import sys

subprocess.check_call([
    sys.executable,
    str(ROOT / "infer" / "track.py"),
    "--weights",
    str(weights),
    "--source",
    str(source),
    "--conf",
    "0.20",
    "--trail",
])
````
