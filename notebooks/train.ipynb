# Train the aircraft detector

Thin wrapper around [`train/train.py`](../train/train.py). The 2023 Colab notebook in `original-data/` still talks about face detection — use this one.

Images are **not** in git. Put YOLO frames under `data/images/` and `data/labels/` first (see [`data/README.md`](../data/README.md)).

```python
import os
from pathlib import Path

os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
os.chdir(ROOT)
ROOT
```

```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")
results = model.train(
    data=str(ROOT / "data" / "data.yaml"),
    imgsz=640,
    epochs=50,
    batch=16,
    name="aircraft-yolov8",
    workers=8,
    cache=True,
    augment=True,
    project=str(ROOT / "runs" / "detect"),
)
results
```

```python
metrics = model.val()
metrics.box.map, metrics.box.map50
```
